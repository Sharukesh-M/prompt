```python
import os
from getpass import getpass
from groq import Groq

api_key = os.getenv("GROQ_API_KEY") or getpass("Groq API key: ")
client = Groq(api_key=api_key)
model = os.getenv("GROQ_MODEL", "openai/gpt-oss-20b")
```

```python
prompt_template = """
You are an educational AI assistant.

Explain the following topic in simple language.
Include:
1. Definition
2. Key points
3. One real-world example

Topic: {topic}
"""
```

```python
example_topic = "Machine Learning"
example_prompt = prompt_template.format(topic=example_topic)

print("Generated prompt example:")
print(example_prompt)
```

```python
topic = input("Enter a topic: ").strip()
if not topic:
    raise ValueError("Enter a topic before building the prompt.")

prompt = prompt_template.format(topic=topic)

print("Generated Prompt:")
print(prompt)
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[{"role": "user", "content": prompt}],
)

answer = response.choices[0].message.content or ""
print("\nAI Response:")
print(answer)
```
